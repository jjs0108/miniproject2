```python
from pathlib import Path
import json
import re
import pandas as pd
from IPython.display import display

root = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if p.name == "miniproject2" and (p / "data/train.csv").is_file()), None)
if root is None:
    root = Path.cwd() / "miniproject2"
if not (root / "data/train.csv").is_file():
    raise FileNotFoundError("miniproject2 또는 note 폴더에서 실행하세요.")
data_dir = root / "data"
train = pd.read_csv(data_dir / "train.csv", encoding="utf-8-sig")
test = pd.read_csv(data_dir / "test.csv", encoding="utf-8-sig")
knct_raw = json.loads((data_dir / "K-NCT_v1.4.json").read_text())["data"]
knct = pd.DataFrame(knct_raw)
display(pd.DataFrame({"파일": ["train.csv", "test.csv", "K-NCT_v1.4.json"],
                      "문장 쌍": [len(train), len(test), len(knct)]}))
```
